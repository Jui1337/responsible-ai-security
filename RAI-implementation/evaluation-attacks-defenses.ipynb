```python
def extract_metrics_with_std(all_results, dataset, model, attack):
    key = f"{dataset}_{model}_{attack}"
    try:
        res = all_results[key]["result"]
        avg_dist = res['average_distances']
        std_dist = res.get('std_distances', {})  # Add std_distances if you store them
        
        return {
            "Attack": attack,
            "Model": model,
            "Attack Success Rate (Mean)": res['attack_success_rates'],
            "Attack Success Rate (Std)": res.get('attack_success_rate_std', None),
            "L0 (Mean)": avg_dist['L0 Distance'],
            "L0 (Std)": std_dist.get('L0 Distance', None),
            "L1 (Mean)": avg_dist['L1 Distance'],
            "L1 (Std)": std_dist.get('L1 Distance', None),
            "L2 (Mean)": avg_dist['L2 Distance'],
            "L2 (Std)": std_dist.get('L2 Distance', None),
            "Linf (Mean)": avg_dist['Linf Distance'],
            "Linf (Std)": std_dist.get('Linf Distance', None),
            "Mahalanobis (Mean)": avg_dist['Mahalanobis Distance'],
            "Mahalanobis (Std)": std_dist.get('Mahalanobis Distance', None),
            "Sensitivity (Mean)": avg_dist['Sensitivity'],
            "Sensitivity (Std)": std_dist.get('Sensitivity', None),
            "Outlier Rate (Mean)": res['outliner_rates'],
            "Outlier Rate (Std)": res.get('outliner_rate_std', None),
            "Epsilon": res['epsilons']
        }
    except KeyError:
        print(f"Missing data for: {key}")
        return None
```
